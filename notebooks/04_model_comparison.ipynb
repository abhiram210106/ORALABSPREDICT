# OralAbsPredict — 04: Model Benchmarking & Multi-Algorithm Comparison
## Evaluating 6 Algorithms under Biological Class Imbalance


```python
import json
import pandas as pd
import matplotlib.pyplot as plt

with open('models/metadata/hia_metadata.json', 'r') as f:
    hia_meta = json.load(f)

df_comp = pd.DataFrame(hia_meta['validation_comparison'])
print('Candidate Model Validation Comparison for HIA:')
print(df_comp[['Model', 'Balanced Accuracy', 'F1-Score', 'ROC-AUC', 'Accuracy']])

# Plot comparison
plt.figure(figsize=(10, 5))
plt.barh(df_comp['Model'], df_comp['Balanced Accuracy'], color='#38bdf8')
plt.xlabel('Balanced Accuracy')
plt.title('HIA Model Comparison (Validation Set)')
plt.xlim(0.5, 1.0)
plt.tight_layout()
plt.show()
```
