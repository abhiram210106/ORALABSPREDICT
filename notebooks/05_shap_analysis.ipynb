# OralAbsPredict — 05: Explainable AI (SHAP Analysis)
## Local Attribution and Global Pharmacophore Identification


```python
import joblib
from src.predictor import OralAbsPredictor
from src.explainability import explain_prediction, get_global_feature_importance

predictor = OralAbsPredictor()
smi = 'CC(=O)Nc1ccc(O)cc1' # Paracetamol

res = predictor.predict_hia(smi)
print('HIA Prediction:', res['label'])
print('HIA Probability High:', res['probability_high'])

# Local SHAP Explanation
exp = explain_prediction(
    model=predictor.hia_model,
    feature_names=predictor.hia_features,
    feature_vector=res['feature_vector'],
    top_n=5
)
print('\nTop Contributing Features (Local SHAP):')
print(exp['explanation_table'])

# Global Feature Importance
df_imp, _ = get_global_feature_importance(predictor.hia_model, predictor.hia_features, top_n=10)
print('\nTop 10 Global Features across all compounds:')
print(df_imp[['clean_name', 'importance']])
```
