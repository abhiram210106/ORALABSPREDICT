# OralAbsPredict — 01: Exploratory Data Analysis (EDA)
## Human Intestinal Absorption (HIA) & Human Oral Bioavailability (HOB)


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_loader import load_dataset

# 1. Load HIA Dataset (Hou et al.)
hia_df, hia_s_col, hia_t_col, hia_meta = load_dataset('data/raw/hia_hou.csv')
print('HIA Dataset Shape:', hia_df.shape)
print('HIA Class Distribution:
', hia_df[hia_t_col].value_counts())

# 2. Load HOB Dataset (Ma et al.)
hob_df, hob_s_col, hob_t_col, hob_meta = load_dataset('data/raw/bioavailability_ma.csv')
print('
HOB Dataset Shape:', hob_df.shape)
print('HOB Class Distribution:
', hob_df[hob_t_col].value_counts())
```

```python
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
hia_df[hia_t_col].value_counts().plot(kind='bar', ax=axes[0], color=['#10b981', '#ef4444'])
axes[0].set_title('HIA Class Distribution (0=Low, 1=High)')
axes[0].set_xlabel('Class Label')
axes[0].set_ylabel('Number of Compounds')

hob_df[hob_t_col].value_counts().plot(kind='bar', ax=axes[1], color=['#10b981', '#ef4444'])
axes[1].set_title('HOB Class Distribution (0=Low, 1=High)')
axes[1].set_xlabel('Class Label')
axes[1].set_ylabel('Number of Compounds')
plt.tight_layout()
plt.show()
```
