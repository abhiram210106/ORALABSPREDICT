## RDKit Descriptors, Mordred Indices, and Morgan Circular Fingerprints


```python
from rdkit import Chem
from src.descriptors import calculate_rdkit_descriptors, calculate_mordred_subset, generate_features, get_descriptor_table_df
from src.fingerprints import calculate_morgan_fingerprint

smi = 'CC(=O)Oc1ccccc1C(=O)O' # Aspirin
mol = Chem.MolFromSmiles(smi)

# 1. RDKit Physicochemical Descriptors
rdkit_desc = calculate_rdkit_descriptors(mol)
print('RDKit Descriptors (18 features):')
for k, v in list(rdkit_desc.items())[:6]:
    print(f'  {k}: {v}')

# 2. 512-bit Morgan Circular Fingerprints (ECFP4)
fp = calculate_morgan_fingerprint(mol, radius=2, n_bits=512)
print(f'\nMorgan Fingerprint Shape: {fp.shape} (On bits: {fp.sum()})')

# 3. Unified Vector Extraction
vec = generate_features(smi, n_bits=512)
print(f'\nTotal Unified Feature Vector Shape: {vec.shape}')
```
